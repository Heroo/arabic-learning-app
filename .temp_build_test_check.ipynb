# Verify project build and tests

This notebook checks whether the current React + Vite project builds successfully and whether the test suite passes.

```python
import os
import json
import subprocess
from pathlib import Path

workspace = Path('/workspaces/arabic-learning-app')
print('Workspace:', workspace)
print('Exists:', workspace.exists())
print('Files:')
for path in sorted(workspace.glob('*')):
    print('-', path.name)
```

```python
package_json = workspace / 'package.json'
print('package.json exists:', package_json.exists())
if package_json.exists():
    print(package_json.read_text())

vite_config = workspace / 'vite.config.js'
print('vite.config.js exists:', vite_config.exists())
```

```python
def run(cmd, cwd=None, timeout=600):
    print('Running:', cmd)
    result = subprocess.run(cmd, shell=True, cwd=cwd, capture_output=True, text=True, timeout=timeout)
    print('Return code:', result.returncode)
    if result.stdout:
        print('stdout:
', result.stdout)
    if result.stderr:
        print('stderr:
', result.stderr)
    return result

run('npm install', cwd=str(workspace), timeout=600)
```

```python
build_result = run('npm run build', cwd=str(workspace), timeout=600)
print('Build success:', build_result.returncode == 0)
```

```python
test_result = run('npm test -- --run', cwd=str(workspace), timeout=600)
print('Tests success:', test_result.returncode == 0)
```
